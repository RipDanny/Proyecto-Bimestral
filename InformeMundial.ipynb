## ABSTRACT 
_______________




El presente proyecto desarrolla una simulación computacional del proceso de colección e intercambio de cromos del Álbum Oficial de la Copa Mundial de la FIFA 2026, conformado por 980 cromos distribuidos aleatoriamente en fundas de 7 unidades. El objetivo principal es determinar la cantidad mínima de fundas adicionales necesarias para completar el álbum dentro de una población cerrada de participantes, considerando la presencia de cromos repetidos y la posibilidad de realizar intercambios entre coleccionistas. Para ello, se implementó un modelo probabilístico basado en distribuciones aleatorias, complementado con algoritmos de intercambio que permiten optimizar la obtención de cromos faltantes mediante la colaboración entre participantes.

La solución propuesta combina herramientas de simulación estadística, análisis probabilístico y visualización gráfica para evaluar el comportamiento del sistema bajo diferentes tamaños de población, desde un único participante hasta grupos de cincuenta coleccionistas. Asimismo, se incorporó el cálculo de probabilidades analíticas y experimentales de completar el álbum, así como una estimación matemática del número de fundas adicionales requeridas después de cada ronda de intercambio. Los resultados obtenidos permiten analizar el impacto del número de participantes en la reducción de costos y en la eficiencia del proceso de completación del álbum, proporcionando una aproximación cuantitativa al problema del coleccionismo colaborativo.

_______________
### OBJETIVOS
_______________

* Desarrollar e implementar un sistema de simulación probabilística que permita analizar el proceso de colección e intercambio de cromos del Álbum Oficial de la Copa Mundial de la FIFA 2026, con el fin de determinar la cantidad de fundas adicionales necesarias para completar el álbum dentro de una población cerrada de participantes

* Diseñar e implementar un algoritmo de intercambio de cromos que permita optimizar la distribución de cromos repetidos entre los participantes, considerando intercambios directos y triangulares dentro de una población cerrada.

* Desarrollar un modelo matemático y probabilístico que estime la probabilidad de completar el álbum en función del número de cromos faltantes y del número de participantes involucrados en el proceso de intercambio.

* Validar experimentalmente los resultados obtenidos mediante simulaciones computacionales repetidas para diferentes tamaños de población, analizando el impacto de los intercambios en la reducción del número de fundas adicionales requeridas para completar el álbum.
