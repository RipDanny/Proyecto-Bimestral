## 2. Marco Teórico

### 2.1 Álbum de Cromos con Problema Probabilístico
La colección de cromos puede modelarse como un experimento aleatorio en el cual cada cromo posee la misma probabilidad de aparecer en una funda. En el caso del Álbum Oficial de la Copa Mundial de la FIFA 2026, existen 980 cromos distintos distribuidos aleatoriamente en fundas de siete unidades. Debido a que la selección de cromos es independiente y equiprobable, es posible analizar el proceso mediante herramientas de probabilidad y estadística.

Sea (M) el numero total de cromos del álbum:
$$M= 980$$

y sea (c) el número de cromos contenidos en una funda:
$$c = 7$$

El número mínimo teórico de fundas necesarias para obtener los 980 cromos sin repeticiones se calcula mediante:
$$N_{min} = \frac{M}{c}$$

por lo que: 
$$N_{min} = \frac{980}{7} = 140$$

Sin embargo, debido a la aparición de cromos repetidos, en la práctica se requiere una cantidad superior de fundas. 

### 2.2 Probabilidad de Aparición de un Cromo
Bajo el supuesto de distribución uniforme, cada cromo tiene la misma probabilidad de ser seleccionado. 

La probabilidad de obtener un cromo específico (i) es: 
$$P(i) = \frac{1}{M} = \frac{1}{980}$$

De forma complementaria, la probabilidad de no obtener dicho cromo en una extracción es:
$$ P(\overline {i}) = 1 - P(i) = 1 - \frac{1}{980} $$

Si realizan (k) extracciones independientes, la probabilidad de que el cromo nunca aparezca es:
$$ \left (1-\frac{1}{980} \right)^k $$

### 2.3 Valor Esperado de Cromos Únicos
Para estimar cuántos cromos distintos obtiene un participante después de abrir una determinada cantidad de fundas, se utiliza el concepto de valor esperado. 
Sea (U) el número de cromos únicos obtenidos después de abrir (k) cromos.
El valor esperado viene dado por:
$$ E[U] = M\left [1-\left(1-\frac{1}{M} \right)^k \right] $$

Para un participante que compra inicialmente 140 fundas:
$$ k = 140\times7= 980 $$

Por tanto:
$$980\left [1-\left(1-\frac{1}{980} \right)^{980} \right]$$

Este resultado permite estimar cuántos cromos distintos posee un participante antes de realizar intercambios.

### 2.4 Problema del Coleccionista de Cupones 
El problema del coleccionista de cupones (Coupon Collector Problem) es un modelo clásico de probabilidad que estudia cuántas muestras son necesarias para obtener todos los elementos de una colección. 
El número esperado de cromos necesarios para completar una colección de (M) elementos está dado por:
$$ H_M = \sum _{i=1}^{M} \frac{1}{i} $$

donde (H_M) es el número armónico:
$$ H_M \approx \ln(M) + \gamma $$

con (\gamma) siendo la constante de Euler-Mascheroni. 
$$ \gama\approx 0.57721 $$

Por lo tanto: 
$$ E(T)\aprox M(\ln(M) + \gamma) $$

Este modelo explica por qué completar un álbum resulta cada vez más difícil a medida que disminuye la cantidad de cromos faltantes.

### 2.5 Intercambio de Cromos en una población Cerrada
Una población cerrada se define como un conjunto de participantes que únicamente pueden intercambiar cromos entre ellos, sin recibir cromos de fuentes externas. 
Si existen (n) participantes, cada uno con una colección obtenida de manera independiente, la probabilidad de que un cromo específico no aparezca en ninguno de ellos es:
$$\left(1-\frac{1}{M} \right)^{nk}$$

Por lo tanto, la probabilidad de que dicho cromo sí exista dentro de la comunidad es:
$$\left(1-\frac{1}{M} \right)^{nk}$$

Esta expresión permite modelar el beneficio de aumentar el número de participantes en la población de intercambio.

### 2.6 Probabilidad Analítica de Completar el Álbum
Sea:
 * (m) el número de cromos faltantes.
 * (n) el número de participantes en la población.
La probabilidad de que un cromo faltante pueda encontrarse dentro de la población se aproxima mediante:
$$ 1-\left(1-\frac{1}{980} \right)^{980(n-1)} $$

Suponiendo independencia entre los cromos faltantes, la probabilidad de completar el álbum mediante intercambios es: 
$$ (P_s) ^m $$

o quivalentemenente:
$$\left[1-\left(1-\frac{1}{980} \right)^{980(n-1)} \right]^m $$

Esta fórmula constituye la base analítica utilizada en la simulación para estimar la probabilidad de completar un álbum según el número de participantes y la cantidad de cromos faltantes.

### 2.7 Simulación Monte Carlo 
La simulación Monte Carlo es una técnica computacional que utiliza números aleatorios para aproximar el comportamiento de sistemas probabilísticos complejos. 
En este proyecto se emplea para:
*Simular la compra de fundas.
*Generar cromos aleatorios.
*Ejecutar rondas de intercambio.
*Estimar experimentalmente la probabilidad de completar el álbum.
*Comparar resultados experimentales con los obtenidos mediante el modelo analítico.
La repetición de múltiples simulaciones permite obtener promedios y reducir la variabilidad inherente a los procesos aleatorios.

### 2.8 Estimación de fundas adicionales 
Después de cada ronda de intercambio, un participante puede conservar aún (f) cromos faltantes. 
El número esperado de fundas adicionales se aproxima mediante:
$$\left \lceil \frac{M}{c} \ln\left(\frac{M}{M-f}\right) \right \rceil $$

Para el álbum del Mundial 2026:
$$\left \lceil 140\ln\left(\frac{980}{980-f} \right) \right \rceil $$

Esta expresión permite estimar la cantidad de fundas requeridas para completar los cromos faltantes considerando el efecto de las repeticiones.

## 3. Metodología

Algoritmos Implementados:
### Algoritmo 1: Generación Aleatoria de Cromos
El sistema implementado utiliza un modelo de generación aleatoria para simular la compra de fundas de cromos. Cada funda contiene siete cromos seleccionados aleatoriamente entre los 980 cromos posibles del álbum. 
La simulación se realiza utilizando una distribución uniforme discreta, donde cada cromo posee la misma probabilidad de aparecer:
$$ P(i) = \frac{1}{980} $$

Para cada participante, el algoritmo genera inicialmente:
$$140\text{ fundas} \times 7 \text{ cromos} = 980 \text{ cromos}$$

La implementación computacional utiliza arreglos numéricos para almacenar la cantidad de veces que aparece cada cromo dentro del álbum de cada participante. 

### Algoritmo 2: Algoritmo de Identificación de Cromos Faltantes y Repetidos
Después de la compra inicial de fundas, el sistema determina automáticamente:
*Cromos faltantes:
Un cromo se considera faltante cuando no aparece ninguna vez en el álbum del participante:
$$album[i] = 0$$

El número total de cromos faltantes se calcula mediante:
$$F = \sum_{i=1}^{980} |album[i]|$$

donde (I) representa una función indicadora.

*Cromos repetidos:
Un cromo se considera repetido cuando aparece más de una vez:
$$album[i] > 1$$

El número total de cromos repetidos se calcula mediante:
$$R = \sum_{i=1}^{980} \max(album[i] - 1, 0)$$

Este procedimiento permite identificar qué cromos pueden utilizarse posteriormente en el algoritmo de intercambio.

### Algoritmo 3: Intercambio de Cromos
El algoritmo de intercambio constituye el núcleo principal del proyecto. Su objetivo es reducir la cantidad de cromos faltantes utilizando los cromos repetidos disponibles dentro de una población cerrada de participantes.
*Población Cerrada:
Todos los intercambios se realizan únicamente entre los participantes que pertenecen al sistema simulado, sin intervención de agentes externos.

*Intercambios Directos:
El sistema verifica si un participante posee un cromo faltante y si otro participante dispone de dicho cromo como repetido. 
Si: 
$$album_j[s] > 1$$

 y simultaneamente:
$$album_i[s] = 0$$

entonces el intercambio se realiza.

*Intercambios Triangulares:
Además de los intercambios directos, el algoritmo implementa intercambios triangulares entre tres participantes distintos.
Este mecanismo permite situaciones como:
$$A\leftarrow B$$ 
$$B\leftarrow C$$
$$C\leftarrow A$$

lo que incrementa significativamente la eficiencia del sistema de intercambio y mejora la probabilidad de completar el álbum.

### Algoritmo 4: Cálculo de Probabilidad Analítica 
El sistema incorpora un modelo matemático para estimar la probabilidad de completar el álbum después de una ronda de intercambios.
Sea:
* (m) el número de cromos faltantes.
* (n) el número de participantes.
La probabilidad de que un cromo faltante exista dentro de la población se calcula mediante:
$$ 1-\left(1-\frac{1}{980} \right)^{980(n-1)} $$

La probabilidad de completar todos los cromos faltantes se aproxima mediante:
$$ (P_s) ^m $$

o equivalentemente:
$$\left[1-\left(1-\frac{1}{980} \right)^{980(n-1)} \right]^m $$

Este algoritmo permite comparar resultados teóricos con los resultados experimentales obtenidos mediante simulación.

### Algoritmo 5: Simulación Monte Carlo 
La simulación Monte Carlo se utiliza para validar experimentalmente el comportamiento probabilístico del sistema.
El procedimiento consiste en repetir múltiples veces el proceso completo de:

Cada simulación genera resultados ligeramente distintos debido a la naturaleza aleatoria del sistema.
El promedio de múltiples ejecuciones permite aproximar el comportamiento esperado del modelo probabilístico.
* Variables Analizadas:

### Algoritmo 6: Estimación de Fundas Adicionales 
Después de cada ronda de intercambio, el sistema calcula cuántas fundas adicionales debe comprar un participante en función de los cromos faltantes restantes. 
El cálculo se basa en el modelo del Problema del Coleccionista de Cupones:
$$\left \lceil \frac{980}{7} \ln\left(\frac{980}{980-f} \right) \right \rceil $$

donde:
* (f): cromos faltantes.
*(980): número total de cromos.
*(7): cromos por funda.
Este algoritmo permite modelar el aumento progresivo de dificultad para completar el álbum a medida que disminuye el número de cromos faltantes.

```python
def analytical_prob(missing, participants):
    if missing == 0:
        return 1.0

    p_sticker = 1 - ((1 - 1/TOTAL) ** (TOTAL * (participants - 1)))

    return p_sticker ** missing
```

```python
def extra_packs_needed(missing):
    if missing == 0:
        return 0

    return int (np.ceil((TOTAL / PER_PACK) * np.log(TOTAL / TOTAL - missing)))
```

#### Función Principal de Simulación:
```python
def simulate_many(n_part, reps):
    results = []
    for i in range(reps):
        result = simulate_one(n_part)
        results.append(result)
    return results
```
